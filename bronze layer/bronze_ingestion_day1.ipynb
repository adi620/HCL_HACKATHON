```python
%sql
CREATE SCHEMA IF NOT EXISTS datastore.bronze
```

```python
df = spark.read.csv(
    "/Volumes/datastore/landing/hcl_hackathon/customers.csv",
    header=True,
    inferSchema=True
)
from pyspark.sql.functions import current_timestamp

#adding created date column for audit log purpose
df = df.withColumn("created_date", current_timestamp())

df.write.format("delta").mode("overwrite").saveAsTable("datastore.bronze.customers")
```

```python
from pyspark.sql.functions import current_timestamp, col

df_product = spark.read.json(
    "/Volumes/datastore/landing/hcl_hackathon/products_json.json"
).filter(col("product_id").isNotNull()).drop("_corrupt_record")

#adding created date column for audit log purpose
df_product = df_product.withColumn("created_date", current_timestamp())

#saving the JSON data as a structured Delta table in the bronze schema
df_product.write.format("delta").mode("overwrite").saveAsTable("datastore.bronze.products")

# df_product.display()
```
