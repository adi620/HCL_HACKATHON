```python
df_day2 = spark.read.csv(
    "/Volumes/datastore/landing/hcl_hackathon/customer_updates_day2.csv",
    header=True,
    inferSchema=True
)
from pyspark.sql.functions import current_timestamp
df_day2 = df_day2.withColumn("created_date", current_timestamp())

df_day2.write.format("delta").mode("append").saveAsTable("datastore.bronze.customers")
```
